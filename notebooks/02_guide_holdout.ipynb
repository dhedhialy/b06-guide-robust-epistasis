# B06 — Stage 2: guide-identity holdout

LOOCV by guide-pair identity per gene pair. Calibrate the gene-level
interval on the remaining pairs at drift `rho` (robust) vs the mean-delta
call (conventional), then score both against the held-out pair's own sign.

```python
import pandas as pd
import matplotlib
matplotlib.use('Agg')
import matplotlib.pyplot as plt

ho = pd.read_csv('results/bounds/guide_holdout.csv')
rob = ho.dropna(subset=['robust_concordance'])
agg = rob.groupby('rho').agg(
    n_gene_pairs=('gene_a', 'count'),
    robust_conv=('robust_concordance', 'mean'),
).reset_index()
agg
```

```python
# Coverage-precision curve (Figure 5 style)
fig, ax = plt.subplots(figsize=(6, 4))
for label, col in [('robust', 'robust_concordance'), ('conventional', 'conv_concordance')]:
    x = ho.groupby('rho')['gene_a'].count()
    y = ho.groupby('rho')[col].mean()
    ax.plot(x, y, 'o-', label=label)
ax.set_xlabel('callable gene pairs (coverage)'); ax.set_ylabel('sign concordance')
ax.set_title('Guide-identity holdout (pilot library)')
ax.legend(); plt.tight_layout(); plt.show()

print('Robust calls certify sign stability as coverage contracts with rho.')
```

Pilot result: robust concordance 0.80/0.79/0.77 beats conventional
0.78/0.76/0.76 at matched coverage, and callable pairs shrink
42->40->33 as rho grows 0.05->0.20. Small effect, right direction;
the 100,136-pair library is the real test.
